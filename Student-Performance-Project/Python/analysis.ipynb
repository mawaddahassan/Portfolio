# Student Performance Analysis Using Python
## Project Objective
This project analyzes student performance data using Python and data visualization techniques to identify trends and factors affecting academic performance
## Tools Used
<ul>
<li>Python</li>
<li>pandas</li>
<li>matplotlib</li>
</ul>
## Dataset
Student performance dataset containing math, reading, and writing scores with demographic information.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("StudentsPerformance.csv")
df.head()
```

```python
# Basic Exploration
df.info()
df.isnull().sum()
```

```python
df.describe()
```

```python
# Creating Average Score
df["average_score"] = (df["math score"] + df["reading score"] + df["writing score"])/3
```

```python
# Gender Performance
gender_avg = df.groupby("gender")["average_score"].mean()
print(gender_avg)
```

```python
# Visualization of Gender Performance
gender_avg.plot(kind="bar")
plt.title("Average Score by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Score")
plt.savefig("gender_chart.png")
```

```python
# Analysis of Nutritions
lunch_avg = df.groupby("lunch")["average_score"].mean()
print(lunch_avg)
```

```python
# Lunch vs. Performance
lunch_avg.plot(kind="bar")

plt.title("Lunch Type vs Performance")
plt.xlabel("Lunch Type")
plt.ylabel("Average Score")
plt.savefig("lunch_chart.png")
```

```python
# Top Performing Students

top_students = df.sort_values(
    by = "average_score",
    ascending=False
)
top_students.head(10)
```

```python
# Visualization of Grades
df["average_score"].plot(kind="hist", bins=10)
plt.xlabel("Distribution of Average Scores")
plt.xlabel("Average Score")

plt.savefig("grades_histogram.png")
```

### Key Finfings
<ol>
<li> Female Students achieved higher scores (69.569498 average of female grades, 65.837483 average of male).</li>
<li>Students with standard lunch plans tended to perform better than students with free/reduced lunch.</li>
